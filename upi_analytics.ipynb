```python
status_counts.plot(
    kind="bar",
    title="Transaction Status Distribution"
)
```

```python
status_percentage = (
    df["transaction_status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(status_percentage)
```

```python
transaction_type_counts = df["transaction_type"].value_counts()

print(transaction_type_counts)
```

```python
transaction_type_counts.plot(
    kind="bar",
    title="Transactions by Transaction Type"
)
```

```python
transaction_type_value = (
    df.groupby("transaction_type")["amount_inr"]
    .sum()
    .sort_values(ascending=False)
)

print(transaction_type_value)
```

```python
merchant_counts = df["merchant_category"].value_counts()

print(merchant_counts)
```

```python
merchant_value = (
    df.groupby("merchant_category")["amount_inr"]
    .sum()
    .sort_values(ascending=False)
)

print(merchant_value)
```

```python
merchant_average = (
    df.groupby("merchant_category")["amount_inr"]
    .mean()
    .sort_values(ascending=False)
)

print(merchant_average)
```

```python
monthly_transactions = (
    df.groupby(["year", "month"])
    .size()
)

print(monthly_transactions)
```

```python
monthly_value = (
    df.groupby(["year", "month"])["amount_inr"]
    .sum()
)

print(monthly_value)
```

```python
monthly_transactions.plot(
    kind="line",
    marker="o",
    title="Monthly Transaction Volume"
)
```

```python
day_counts = df["day_name"].value_counts()

print(day_counts)
```

```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_counts = (
    df["day_name"]
    .value_counts()
    .reindex(day_order)
)

print(day_counts)
```

```python
day_counts.plot(
    kind="bar",
    title="Transactions by Day of Week"
)
```

```python
hourly_transactions = (
    df.groupby("hour_of_day")
    .size()
)

print(hourly_transactions)
```

```python
hourly_transactions.plot(
    kind="line",
    marker="o",
    title="Transactions by Hour"
)
```

```python
weekend_analysis = (
    df.groupby("is_weekend")
    .agg(
        transactions=("transaction_id", "count"),
        total_value=("amount_inr", "sum"),
        average_amount=("amount_inr", "mean")
    )
)

print(weekend_analysis)
```

```python
state_transactions = (
    df["sender_state"]
    .value_counts()
)

print(state_transactions.head(10))
```

```python
state_value = (
    df.groupby("sender_state")["amount_inr"]
    .sum()
    .sort_values(ascending=False)
)

print(state_value.head(10))
```

```python
bank_transactions = (
    df["sender_bank"]
    .value_counts()
)

print(bank_transactions)
```

```python
bank_value = (
    df.groupby("sender_bank")["amount_inr"]
    .sum()
    .sort_values(ascending=False)
)

print(bank_value)
```

```python
bank_success = (
    df.assign(
        successful=df["transaction_status"] == "SUCCESS"
    )
    .groupby("sender_bank")["successful"]
    .mean()
    .mul(100)
    .round(2)
)

print(bank_success.sort_values(ascending=False))
```

```python
device_analysis = (
    df.groupby("device_type")
    .agg(
        transactions=("transaction_id", "count"),
        total_value=("amount_inr", "sum"),
        average_amount=("amount_inr", "mean")
    )
    .sort_values("transactions", ascending=False)
)

print(device_analysis)
```

```python
network_analysis = (
    df.groupby("network_type")
    .agg(
        transactions=("transaction_id", "count"),
        total_value=("amount_inr", "sum")
    )
)

print(network_analysis)
```

```python
fraud_percentage = (
    df["fraud_flag"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(fraud_percentage)
```

```python
fraud_analysis = (
    df.groupby("fraud_flag")
    .agg(
        transactions=("transaction_id", "count"),
        total_value=("amount_inr", "sum"),
        average_amount=("amount_inr", "mean")
    )
)

print(fraud_analysis)
```

```python
fraud_by_type = (
    df.groupby("transaction_type")
    .agg(
        transactions=("transaction_id", "count"),
        fraud_transactions=("fraud_flag", "sum")
    )
)

fraud_by_type["fraud_rate"] = (
    fraud_by_type["fraud_transactions"]
    / fraud_by_type["transactions"]
    * 100
).round(2)

fraud_by_type
```

```python
fraud_by_device = (
    df.groupby("device_type")
    .agg(
        transactions=("transaction_id", "count"),
        fraud_transactions=("fraud_flag", "sum")
    )
)

fraud_by_device["fraud_rate"] = (
    fraud_by_device["fraud_transactions"]
    / fraud_by_device["transactions"]
    * 100
).round(2)

fraud_by_device
```

```python
fraud_by_hour = (
    df.groupby("hour_of_day")
    .agg(
        transactions=("transaction_id", "count"),
        fraud_transactions=("fraud_flag", "sum")
    )
)

fraud_by_hour["fraud_rate"] = (
    fraud_by_hour["fraud_transactions"]
    / fraud_by_hour["transactions"]
    * 100
).round(2)

fraud_by_hour
```

```python
fraud_by_bank = (
    df.groupby("sender_bank")
    .agg(
        transactions=("transaction_id", "count"),
        fraud_transactions=("fraud_flag", "sum")
    )
)

fraud_by_bank["fraud_rate"] = (
    fraud_by_bank["fraud_transactions"]
    / fraud_by_bank["transactions"]
    * 100
).round(2)

fraud_by_bank.sort_values(
    "fraud_rate",
    ascending=False
)
```
